```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Set parameters
sample_size = 30
confidence_level = 95

# Calculate the critical value for the Z-score
critical_value = stats.norm.ppf((1 + confidence_level / 100) / 2)

# Generate an array of population standard deviations
population_std_devs = np.arange(1, 101)

# Calculate the margin of error for each population standard deviation
margin_of_errors = critical_value * (population_std_devs / np.sqrt(sample_size))

# Plot the relationship
plt.plot(population_std_devs, margin_of_errors)
plt.xlabel("Population Standard Deviation")
plt.ylabel("Margin of Error")
plt.title("Relationship Between Margin of Error and Population Standard Deviation (Z Procedure)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Set parameters
population_std_dev = 20
confidence_level = 95

# Calculate the critical value for the Z-score
critical_value = stats.norm.ppf((1 + confidence_level / 100) / 2)

# Generate an array of sample sizes
sample_sizes = np.arange(2, 101)

# Calculate the margin of error for each sample size
margin_of_errors = critical_value * (population_std_dev / np.sqrt(sample_sizes))

# Plot the relationship
plt.plot(sample_sizes, margin_of_errors)
plt.xlabel("Sample Size")
plt.ylabel("Margin of Error")
plt.title("Relationship Between Margin of Error and Sample Size (Z Procedure)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Set parameters
sample_size = 30
population_std_dev = 20

# Generate an array of confidence levels from 50% to 99%
confidence_levels = np.arange(50, 100)

# Calculate the critical values (Z-scores) for each confidence level
critical_values = stats.norm.ppf((1 + confidence_levels / 100) / 2)

# Calculate the margin of error for each critical value
margin_of_errors = critical_values * (population_std_dev / np.sqrt(sample_size))

# Plot the relationship
plt.plot(confidence_levels, margin_of_errors)
plt.xlabel("Confidence Level (%)")
plt.ylabel("Margin of Error")
plt.title("Relationship Between Margin of Error and Critical Value (Z Procedure)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
test_df = pd.read_csv('test.csv')
train_df = pd.read_csv('train.csv')
```

```python
df = pd.concat([test_df,train_df.drop(columns='Survived')])
df
```

```python
sns.kdeplot(df['Fare'])
```

```python
plt.hist(df['Fare'],density=True,edgecolor='black')
plt.show()
```

```python
df['Fare'].skew()
```

```python
# sample size = 30 and 10 times 
df = df.sample(1309)
df
```

```python
sample = []
for i in range(10):
    sample.append(df['Fare'].sample(30).values.tolist())

sample    
```

```python
sample = np.array(sample)
# sample
```

```python
sampling_mean = sample.mean(axis=1)
sampling_mean
```

```python
sampling_std = sample.std(axis=1)
sampling_std
```

```python
# calculate s : which is mean of sampling std
sample_std = np.mean(sampling_std)
sample_std
```

```python
# calculate x-bar : which is mean of all sampling mean
sample_mean = np.mean(sampling_mean)
sample_mean
```

```python
# t alpha/2 : from t table value and 95% at x axis and y is degree of freedom or no. of samples = 30
t = 2.042
```

```python
min_limit = sample_mean - t * (sample_std/np.sqrt(30))
max_limit = sample_mean + t * (sample_std/np.sqrt(30))
```

```python
# confidence interval for avg fare with 95% confidence level is:
print(f"95% confidence interval: ({min_limit:.2f}, {max_limit:.2f})")
```

```python
# confidence interval for avg fare with 50% confidence level is:

min_limit = sample_mean - 0.683 * (sample_std/np.sqrt(30))
max_limit = sample_mean + 0.683 * (sample_std/np.sqrt(30))
print(f"50% confidence interval: ({min_limit:.2f}, {max_limit:.2f})")
```
