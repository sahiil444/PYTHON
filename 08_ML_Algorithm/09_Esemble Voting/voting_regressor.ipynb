```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("altavish/boston-housing-dataset")

print("Path to dataset files:", path)
```

```python
import os
print(os.listdir(path))
```

```python
df = pd.read_csv(os.path.join(path, 'HousingData.csv'))
df
```

```python
df = df.fillna(df.mean(numeric_only=True))
```

```python
X = df.drop(columns='MEDV').values
y = df['MEDV'].values
```

```python
X.shape
```

```python
y.shape
```

```python
X
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

```python
estimators = [('lr',lr),('dt',dt),('svr',svr)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```

```python
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
```

```python
model_vc = VotingRegressor(estimators=estimators,weights=[2,1,1])
x = cross_val_score(
    estimator=model_vc,
    X = X,
    y = y,
    cv=10,
    scoring='r2'
)
print(np.round(np.mean(x),2))
```

```python
# using the same algorithm

dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```

```python
Score = []

for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            for l in range(1,4):
                for m in range(1,4):

                    vr = VotingRegressor(
                        estimators,
                        weights=[i,j,k,l,m]
                    )

                    scores = cross_val_score(
                        vr, X, y,
                        scoring='r2',
                        cv=10
                    )

                    Score.append([
                        i,
                        j,
                        k,
                        l,
                        m,
                        np.round(np.mean(scores), 2)
                    ])
```

```python
cross_val_value_data = pd.DataFrame(Score,columns=['i','j','k','l','m','score'])
```
